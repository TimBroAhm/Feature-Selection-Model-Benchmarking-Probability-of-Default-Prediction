# netPay() Function (Payroll System)

```
def calculate_salary(basic_salary):
    # Employee pension contribution (7%)
    pension = basic_salary * 0.07

    taxable_income = basic_salary

    # Ethiopian monthly income tax calculation
    if taxable_income <= 2000:
        tax = 0

    elif taxable_income <= 4000:
        tax = (taxable_income * 0.15) - 300
    elif taxable_income <= 7000:
        tax = (taxable_income * 0.20) - 500
    elif taxable_income <= 10000:
        tax = (taxable_income * 0.25) - 850
    elif taxable_income <= 14000:
        tax = (taxable_income * 0.30) - 1350
    else:
        tax = (taxable_income * 0.35) - 2050

    net_pay = basic_salary - tax - pension

    print(f"Basic Salary : {basic_salary:.2f}")
    print(f"Tax          : {tax:.2f}")
    print(f"Pension (7%) : {pension:.2f}")
    print(f"Net Pay      : {net_pay:.2f}")

    return {
        "basic_salary": basic_salary,
        "tax": tax,
        "pension": pension,
        "net_pay": net_pay
    }


# Example
calculate_salary(10000)
```
